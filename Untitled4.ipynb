```python
#set operation
#union
python_students = {"anuradha","sarang","sidhi","yashi","shiva"}
java_students={"siddharth","dhanu","monecca"}
python_students.union(java_students)

```

```python
#intersection
python_students = {"anuradha","sarang","sidhi","yashi","shiva"}
java_students={"siddharth","dhanu","monecca","shiva"}
python_students.intersection(java_students)
```

```python
#difference of set
python_students = {"anuradha","sarang","sidhi","yashi","shiva"}
java_students={"siddharth","dhanu","monecca"}
python_students.difference(java_students)
```

```python
#symmetric difference
python_students = {"anuradha","sarang","sidhi","yashi","shiva"}
java_students={"siddharth","dhanu","monecca"}
python_students.symmetric_difference(java_students)
```
